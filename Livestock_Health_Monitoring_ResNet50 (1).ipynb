```python
from PIL import Image
import os

def check_images(folder):

    valid = 0
    corrupted = []

    image_files = [
        file for file in os.listdir(folder)
        if file.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    for filename in image_files:

        path = os.path.join(folder, filename)

        try:
            with Image.open(path) as img:
                img.verify()

            valid += 1

        except Exception:
            corrupted.append(filename)

    return valid, corrupted


healthy_valid, healthy_corrupted = check_images(
    os.path.join(DATASET_DIR, "healthy")
)

lsd_valid, lsd_corrupted = check_images(
    os.path.join(DATASET_DIR, "lumpy_skin_disease")
)

print("Healthy:")
print("Valid:", healthy_valid)
print("Corrupted:", len(healthy_corrupted))

print("\nLumpy Skin Disease:")
print("Valid:", lsd_valid)
print("Corrupted:", len(lsd_corrupted))
```

```python
from PIL import Image, ImageDraw, ImageFont
import os
import math

def create_contact_sheet(folder, start=0, count=50, columns=5):

    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    selected = files[start:start + count]

    thumb_width = 180
    thumb_height = 150
    label_height = 25

    rows = math.ceil(len(selected) / columns)

    sheet = Image.new(
        "RGB",
        (
            columns * thumb_width,
            rows * (thumb_height + label_height)
        ),
        "white"
    )

    draw = ImageDraw.Draw(sheet)

    for i, filename in enumerate(selected):

        path = os.path.join(folder, filename)

        try:
            image = Image.open(path).convert("RGB")
            image.thumbnail((thumb_width - 10, thumb_height - 10))

            x = (i % columns) * thumb_width
            y = (i // columns) * (thumb_height + label_height)

            image_x = x + (thumb_width - image.width) // 2
            image_y = y + 5

            sheet.paste(image, (image_x, image_y))

            draw.text(
                (x + 5, y + thumb_height),
                filename[:24],
                fill="black"
            )

        except Exception:
            pass

    display(sheet)
```

```python
import os
import shutil

SOURCE = "/content/drive/MyDrive/Livestock_Health_Monitoring/dataset"
CLEAN = "/content/drive/MyDrive/Livestock_Health_Monitoring/cleaned_dataset"

os.makedirs(CLEAN, exist_ok=True)

for folder in ["healthy", "lumpy_skin_disease"]:
    os.makedirs(os.path.join(CLEAN, folder), exist_ok=True)

    source_folder = os.path.join(SOURCE, folder)
    clean_folder = os.path.join(CLEAN, folder)

    for file in os.listdir(source_folder):
        if file.lower().endswith((".jpg", ".jpeg", ".png")):
            shutil.copy2(
                os.path.join(source_folder, file),
                os.path.join(clean_folder, file)
            )

print("Cleaned dataset ready!")
```

```python
import os
import shutil
import random

CLEAN = "/content/drive/MyDrive/Livestock_Health_Monitoring/cleaned_dataset"
SPLIT = "/content/drive/MyDrive/Livestock_Health_Monitoring/split_dataset"

random.seed(42)

classes = ["healthy", "lumpy_skin_disease"]

for split in ["train", "val", "test"]:
    for c in classes:
        os.makedirs(
            os.path.join(SPLIT, split, c),
            exist_ok=True
        )

for c in classes:

    folder = os.path.join(CLEAN, c)

    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(files)

    n = len(files)

    train_files = files[:int(0.70 * n)]
    val_files = files[int(0.70 * n):int(0.85 * n)]
    test_files = files[int(0.85 * n):]

    for f in train_files:
        shutil.copy2(
            os.path.join(folder, f),
            os.path.join(SPLIT, "train", c, f)
        )

    for f in val_files:
        shutil.copy2(
            os.path.join(folder, f),
            os.path.join(SPLIT, "val", c, f)
        )

    for f in test_files:
        shutil.copy2(
            os.path.join(folder, f),
            os.path.join(SPLIT, "test", c, f)
        )

    print(
        c,
        "-> Train:", len(train_files),
        "Val:", len(val_files),
        "Test:", len(test_files)
    )

print("\nDataset split completed!")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_dir = os.path.join(SPLIT, "train")
val_dir = os.path.join(SPLIT, "val")
test_dir = os.path.join(SPLIT, "test")

train_data = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True,
    seed=42
)

val_data = tf.keras.utils.image_dataset_from_directory(
    val_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

test_data = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("Classes:", train_data.class_names)
print("Data loading complete!")
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers, models

# ResNet50 base model
base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Base model ko initially freeze rakho
base_model.trainable = False

# New classification layers
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(1, activation="sigmoid")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Model summary
model.summary()
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

SPLIT_DIR = "/content/drive/MyDrive/Livestock_Health_Monitoring/split_dataset"

train_ds = tf.keras.utils.image_dataset_from_directory(
    SPLIT_DIR + "/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    SPLIT_DIR + "/val",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    SPLIT_DIR + "/test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("Classes:", train_ds.class_names)
print("Data loading complete!")
```

```python
EPOCHS = 10

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)
```

```python
# STEP 6: Evaluate on Test Dataset

test_loss, test_accuracy = model.evaluate(test_ds)

print("Test Accuracy:", f"{test_accuracy * 100:.2f}%")
print("Test Loss:", f"{test_loss:.4f}")
```

```python
# STEP 7: Confusion Matrix and Classification Report

import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# True labels
y_true = np.concatenate([y for x, y in test_ds], axis=0).astype(int).flatten()

# Model predictions
y_prob = model.predict(test_ds, verbose=0).flatten()
y_pred = (y_prob >= 0.5).astype(int)

# Class names
class_names = test_ds.class_names

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)

print("Confusion Matrix:")
print(cm)

# Classification Report
print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    digits=4
))

# Plot Confusion Matrix
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - ResNet50")
plt.show()
```
